```python
import pandas as pd
```

```python
notas = pd.read_csv("https://raw.githubusercontent.com/alura-cursos/data-science-analise-exploratoria/main/Aula_0/ml-latest-small/ratings.csv")
notas
```

```python
notas.columns = ["filmID", "nota","momento", "momento_generico"]
notas.head()
```

```python
notas ["nota"].unique()
```

```python
notas ["nota"].value_counts
```

```python
notas ["nota"].plot(kind="hist")
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
médias_por_filmes = notas.groupby('filmID')['nota'].mean()
```

```python
sns.histplot(médias_por_filmes)
```

```python
médias_por_filmes.describe()
```

```python
import matplotlib.pyplot as plt
sns.displot(médias_por_filmes,kde=True)
plt.hist("Histograma das médias dos filmes")

```

```python
filmes = pd.read_csv("https://raw.githubusercontent.com/alura-cursos/data-science-analise-exploratoria/main/Aula_0/ml-latest-small/movies.csv")
filmes.columns = ["filmeId", "titulo", "generos"]
filmes.head()
```

```python
notas.query("filmID==1")["nota"].mean() # Média das notas de Toy Story
notas.query("filmID==2")["nota"].mean()         # Média das notas de Jumanji
```

```python
medias_por_filme = notas.groupby("filmID")["nota"].mean()
medias_por_filme.head()
```

```python
filmes_com_media = pd.merge(filmes, medias_por_filme, left_on='filmeId', right_index=True)
display(filmes_com_media.head())
```
